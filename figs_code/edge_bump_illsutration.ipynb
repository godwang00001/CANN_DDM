```python
import matplotlib.pyplot as plt
import numpy as np

```

```python
# Generate Gaussian curve data
x = np.linspace(-10, 10, 400)
y = np.exp(-x**2 / 4) / np.sqrt(2 * np.pi)

# Plot with updated custom color and thicker line
plt.figure(figsize=(6,4))
plt.plot(x, y, color="#716db2", linewidth=16)
plt.savefig("bump.svg", format="svg", transparent=True, bbox_inches="tight", metadata=None)
plt.axis("off")  # remove axes

# Save as PNG with transparent background
```

```python

# Generate Gaussian curve data
x = np.linspace(-12, 12, 400)
y = np.exp(-0.5*np.exp(0.5*x))

# Plot with updated custom color and thicker line
plt.figure(figsize=(6,4))
plt.plot(x, y, color="#1d52a1", linewidth=16)
plt.axis("off")  # remove axes

# Save as PNG with transparent background
```

```python

# Generate Gaussian curve data
x = np.linspace(-4, 4, 400)
y = np.exp(-0.5*np.exp(2*x))

# Plot with updated custom color and thicker line
plt.figure(figsize=(6,4))
plt.plot(x[1:], -np.diff(y), color="#1d52a1", linewidth=16)
plt.axis("off")  # r
```
