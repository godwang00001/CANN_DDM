```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(1, 1000, 1000)
y1 = np.exp(-0.1*np.exp(.01*(x-500)))
y2 = np.exp(-0.3*np.exp(.01*(x-500)))
y3 = np.exp(-0.5*np.exp(.01*(x-500)))
y4 = np.exp(-0.7*np.exp(.01*(x-500)))
fig, ax = plt.subplots(1,1, figsize=(6,4))
ax.plot(x, y1, color="#1d52a1", alpha=0.9, linewidth=2, label=r"$x=0.9$")
ax.plot(x, y2, color="#1d52a1", alpha=0.7, linewidth=2, label=r"$x=0.7$")
ax.plot(x, y3, color="#1d52a1", alpha=0.5, linewidth=2, label=r"$x=0.5$")
ax.plot(x, y4, color="#1d52a1", alpha=0.3, linewidth=2, label=r"$x=0.3$")
ax.set_xticks([0, 1000], [0, 1000],fontsize=12)
ax.set_yticks([0, 1], [0, 1], fontsize=12)
ax.legend(fontsize=12,frameon=False)
ax.set_xlabel("Edge neuron index", fontsize=16)
ax.set_ylabel("Firing rate", fontsize=16)

#plt.axis("off")  # remove axes
```

```python
c = 0.15
mu_all = np.logspace(-1, 1, 10)
color_bump = "#716db2"
sigma_all = c * mu_all
f_tc = lambda x, mu, sigma: np.exp(-0.5*((x-mu)**2)/sigma**2)

x = np.linspace(0, 10, 5000)
fig, ax = plt.subplots(2, 1, figsize=(6, 4), gridspec_kw={'hspace': 0.25})
for i, (mu, sigma) in enumerate(zip(mu_all, sigma_all)):
    ax[0].plot(x, f_tc(x, mu, sigma), color=color_bump, alpha=1 - 0.8 * i / len(mu_all), linewidth=2)
ax[0].set_xlabel("Time (s)", fontsize=14)
#ax[0].set_ylabel("Firing rate", fontsize=14)
ax[0].set_yticks([0, 1])

ax[1].semilogx()
for i, (mu, sigma) in enumerate(zip(mu_all, sigma_all)):
    ax[1].plot(x, f_tc(x, mu, sigma), color=color_bump, alpha=1 - 0.8 * i / len(mu_all), linewidth=2)
ax[1].set_xlabel("Time (s)", fontsize=14)
#ax[1].set_ylabel("Firing rate", fontsize=14)
ax[1].set_yticks([0, 1])
ax[1].set_xlim(0.05, 10)
```

```python
c = 0.1
mu_all = np.logspace(-0.5, 0.7, 10)
s_all = mu_all[::-1]
color_bump = "#716db2"
color_edge = "#1d52a1"
sigma_all = c * mu_all
f_tc = lambda x, mu, sigma: np.exp(-0.5 * ((x - mu) ** 2) / sigma ** 2)
f_tcc = lambda x, s: np.exp(-s * x)

x = np.linspace(0, 8, 5000)
fig, ax = plt.subplots(1, 2, figsize=(10, 3))

# Top-left: edge kernels linear scale
for i, mu in enumerate(mu_all):
    ax[0].plot(
        x, f_tcc(x, mu), color=color_edge,
        alpha=1 - 0.8 * i / len(mu_all), linewidth=2
    )
ax[0].set_xlabel("Time (s)", fontsize=16)
ax[0].set_yticks([0, 1])
ax[0].set_xlim(0.05, 10)

# Top-right: edge kernels log scale
ax[1].semilogx()
for i, s in enumerate(s_all):
    ax[1].plot(
        x, f_tcc(x, s), color=color_edge,
        alpha=1 - 0.8 * i / len(mu_all), linewidth=2
    )
ax[1].set_xlabel("Time (s)", fontsize=16)
ax[1].set_yticks([0, 1])
ax[1].set_xlim(0.05, 10)

for a in ax.flatten():
    a.tick_params(axis='both', which='major', labelsize=16)
    for spine in a.spines.values():
        spine.set_visible(False)
        


```
